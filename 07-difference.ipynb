# CIP203 - Maximizing GPU usage with MIGs, MPS, and Time-Slicing: 
## Conclusion: What is the right option for me ? 

**Questions**
* How do I know whether I need CUDA streams, MIGs, MPS, or Time-Slicing ?

**Objectives**
* Get to know the difference between GPU sharing technologies
* Learn how to choose wisely one technology or another

### Features of the GPU sharing technologies <br>
Lets compare the features of CUDA Streams, MIGs, MPS, and Time-Slicing

![alt text](./images/table-streams.vs.MIG.vs.MPS.vs.Time-Slicing.png "Title")

### Why choose CUDA Streams ?<br>
* If you are confident with programming CUDA streams
* If you have a single process GPU optimization
* If you are manging multiple tasks within a single application
* If you want to "hide" a CPU-GPU communiction overhead<br>

**Pros:**
- improved GPU utilization
- increased concurrency and throughput
- asynchronous operations, i.e. allowing the CPU to continue processing while the GPU is busy
- prioritization, i.e. allowing the GPU scheduler to prioritize kernels from high-priority streams over those in lower-priority streams<br>

**Cons:**
- increased complexity
- it requires you to re-write your code
- potential for resource contention
- limited isolation
- overhead
- scheduling challenges

### Why choose MIGS ?<br>
* If jobs vary wildly in size or you’ve got strict multi-tenancy
* if your jobs require full instance isolation (both memory and SMs)
* if modifying your code for better performance is somewhat difficult<br>

**Pros:**
- improved GPU utilization
- Hardware-Level Isolation, i.e. each MIG instance has dedicated compute cores, memory, and cache
- Predictable Performance, i.e. the dedicated resources ensure consistent latency and throughput
- Enhanced Security, i.e. the workloads and data from different users are kept secure
- Fault Containment, i.e. a crash in one MIG instance typically will not affect the entire physical GPU
  
**Cons:**
  - Limited Hardware Support : only A100 and H100s supported, older generations are not
  - Static Partitioning: cnce a GPU is partitioned into MIG instances, the configuration is static and cannot be dynamically adjusted
  - Performance Overhead: at least 1/7 of the CUDA cores are reserved (thus wasted for us)
  - Limited Inter-Instance Communication

### Why choose MPS ?<br>
* if you need to run multiple instances of the same application but each instance is too small to saturate GPU
* If memory and fault isolation are not critical<br>

**Pros:** 
- high GPU utilization
- higher overall performance
- better memory management
- enhanced scalability<br>

**Cons:**
- avoid exceeding the available GPU memory
- it requires balancing the number of processes and available GPU memory to prevent performance issues
- it does not affect the concurrency of non-GPU operations

### Why choose Time-Slicing ?<br>
* If you have older GPUs that don't support MIG
* If memory and fault isolation are not critical
* If you need to maximize the number of users/processes sharing a GPU<br>

**Pros:** 
- high GPU utilization
- higher overall performance
- better memory management
- enhanced scalability<br>

**Cons:**
- avoid exceeding the available GPU memory
- it requires balancing the number of processes and available GPU memory to prevent performance issues
- it does not affect the concurrency of non-GPU operations
